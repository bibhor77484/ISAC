```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sionna.rt import load_scene, Transmitter, Receiver, PlanarArray
from sionna.rt.path_solvers import PathSolver
from sionna.rt import *
import mitsuba as mi
import sys

np.random.seed(100)

np.set_printoptions(suppress=True)
np.set_printoptions(threshold=sys.maxsize)
np.set_printoptions(threshold=np.inf, linewidth=np.nan)
```

```python
# This function checks if the point is on the surface of the target or not. 
# To do that, we calculate the perpendicular distance of the point w.r.t. center of target from all the surface. 
def check_inside(point, obj_pos, obj_ori, obj_dim):
    theta = obj_ori[0] # Only rotation is along the z-axis
    
    changed_x = np.array([np.cos(theta), np.sin(theta), 0])
    changed_y = np.array([-np.sin(theta), np.cos(theta), 0])
    changed_z = np.array([0, 0, 1])
    
    vec2point = point - obj_pos
    
    x_dist = np.abs(np.sum(vec2point*changed_x))
    y_dist = np.abs(np.sum(vec2point*changed_y))
    z_dist = np.abs(np.sum(vec2point*changed_z))
    
    if ((x_dist < obj_dim[0]/2+0.01) and (y_dist < obj_dim[1]/2+0.01) and (z_dist < obj_dim[2]/2+0.01)):
        return 1
    else:
        return 0
```

```python
### Generating Trajectory 1 ###

# Locations of the Transmitter and receiver
rx_locations = np.array([-13, 20, 1]) 
tx_locations = np.array([-13, 0, 1])

obj_pos = np.array([100, 200, 0.9]) # Initial position of the object in static environment
obj_pos_dynamic = np.array([-9, 9, 0.9]) # Initial position of the object in static environment
obj_dim = np.array([0.25, 0.5, 1.8]) # Object dimension
obj_ori = np.array([0, 0, 0])

K = 500 # No. of times the simulation will run
Te = 0.02 # Seconds between successive simulation runs
obj_vel = np.array([1, 0.25, 0]) # Speed of the object in m/s

f_c = 3.6192e9
c = 299792458
max_depth = 4
L_paths = 100

# Carrier frequency
scene.frequency = f_c

parameters = np.zeros([K, 4, L_paths]) # Save the channel parameters in a numpy array for being used by ISAC simulator
positions = np.zeros([K, 2]) # Save the position of the object
obj_params = np.zeros([K, 4]) # Save the true AoA, ToF and visibility of the object to the transmitter
detect_snr_thres = 35 # This is the largest ratio of LoS to target path power that makes it a GT detection

# Load the Sionna scene 
scene = load_scene("/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Blender/Outdoor.xml", merge_shapes=False)

# Configure antenna array for all transmitters
scene.tx_array = PlanarArray(num_rows=1,
                             num_cols=1,
                             vertical_spacing=0.5,
                             horizontal_spacing=0.5,
                             pattern="iso",
                             polarization="V")

# Configure antenna array for all receivers
scene.rx_array = PlanarArray(num_rows=1,
                             num_cols=1,
                             vertical_spacing=0.5,
                             horizontal_spacing=0.5,
                             pattern="iso",
                             polarization="VH")

tx = Transmitter(name="tx",
                 position=tx_locations,
                 display_radius=3)

scene.add(tx)

rx = Receiver(name="rx",
              position=rx_locations,
              display_radius=3)

scene.add(rx)

# Instantiate a path solver
# The same path solver can be used with multiple scenes
p_solver  = PathSolver()

scene.get('elm__4').position = mi.Point3f(obj_pos)
scene.get('elm__4').orientation = mi.Point3f(obj_ori)
scene.get('elm__4').radio_material.scattering_coefficient = 0.2


for k in range(0, K):
    
    if k == 30:
        obj_pos = obj_pos_dynamic
    
    # Compute propagation paths
    paths = p_solver(scene,
                     max_depth=max_depth,
                     #max_num_paths_per_src=3,
                     los=True,
                     specular_reflection=True,
                     diffuse_reflection=True,
                     diffraction=True,
                     edge_diffraction=False,
                     refraction=True,
                     synthetic_array=False,
                     seed=41)
    
    poi = np.array(paths.vertices[:, 0, 0, 0, :, :]) # Point of interactions
    toi = np.array(paths.interactions[:, 0, 0, 0, 0, :]) # Type of interactions
    L = np.shape(poi)[2]
    a_v = np.array(paths._a_real[0,0,0,0,:]) + 1j*np.array(paths._a_imag[0,0,0,0,:])
    a_h = np.array(paths._a_real[0,1,0,0,:]) + 1j*np.array(paths._a_imag[0,1,0,0,:])
    a_paths = np.abs(a_v)**2 + np.abs(a_h)**2

    sort_path_idx = np.argsort(a_paths)[::-1]
    los_power = np.max(a_paths)
    i_path = 0
    
    for l in sort_path_idx:
        Doppler = 0
        if (paths.phi_r[0,0,0,0,int(l)] < 0) and (i_path < L_paths): # We are ignoring all the paths that are received from behind of the BS array and also ignoring any path after the strongest 'L_paths'
            
            for order in range(0, max_depth): # Check across the orders of interactions for interactions with the target
                if check_inside(poi[order,0,int(l),:], obj_pos, obj_ori, obj_dim):
                    obj_point = np.array(poi[order, 0, int(l), :]) # Use this as the point of interaction when calculating the Doppler
                    
                    if (order == 0) and toi[1, int(l)] == 0: #(poi[1,0,l,0] == 0 and poi[1,0,l,1] == 0 and poi[1,0,l,2] == 0): # This cannot be used because for some reason, vertices[] array is filled with non-zero coordinates even when interactions is zero for that order of that path.
                        Doppler = np.dot(obj_vel, tx_locations - obj_point)/np.linalg.norm(tx_locations - obj_point)*f_c/c + \
                                  np.dot(obj_vel, rx_locations - obj_point)/np.linalg.norm(rx_locations - obj_point)*f_c/c  # Doppler shift due to the relative motion of the object w.r.t. the BS and UE
                        
                        if a_paths[l]/los_power >= 10**(-0.1*detect_snr_thres):
                            obj_params[k, 0] = 180/np.pi*np.atan2(obj_pos[1]-rx_locations[1], obj_pos[0]-rx_locations[0]) # Save the AoA from the object of interest
                            obj_params[k, 1] = (np.linalg.norm(rx_locations[0:2]-obj_pos[0:2]) + np.linalg.norm(tx_locations[0:2]-obj_pos[0:2]))/c*1e9 # Save the delay from the object of interest
                            obj_params[k, 2] = np.dot(obj_vel, tx_locations - obj_pos)/np.linalg.norm(tx_locations - obj_pos)*f_c/c + \
                                               np.dot(obj_vel, rx_locations - obj_pos)/np.linalg.norm(rx_locations - obj_pos)*f_c/c
                            obj_params[k, 3] = 1 # Save the direct visibility of the object
                            print("{}th path of {}th iteration is a detectable first order reflection from the target".format(int(l), k))
                        else:
                            print("{}th path of {}th iteration is an undetectable first order reflection from the target".format(int(l), k))

                    else:
                        if order == 0:
                            first_end = tx_locations
                            second_end = np.array(poi[1, 0, int(l), :])
                        elif order == max_depth-1:
                            first_end = np.array(poi[max_depth-2, 0, int(l), :])
                            second_end = rx_locations
                        else:
                            first_end = np.array(poi[order-1, 0, int(l), :])
                            second_end = np.array(poi[order+1, 0, int(l), :])

                        Doppler = np.dot(obj_vel, first_end - obj_point)/np.linalg.norm(first_end - obj_point)*f_c/c + \
                                  np.dot(obj_vel, second_end - obj_point)/np.linalg.norm(second_end - obj_point)*f_c/c  # Doppler shift due to the relative motion of the object w.r.t. the environment
                        
                        if a_paths[l]/los_power >= 10**(-0.1*detect_snr_thres):
                            obj_params[k, 0] = 180/np.pi*np.atan2(obj_pos[1]-rx_locations[1], obj_pos[0]-rx_locations[0]) # Save the AoA from the object of interest
                            obj_params[k, 1] = (np.linalg.norm(rx_locations[0:2]-obj_pos[0:2]) + np.linalg.norm(tx_locations[0:2]-obj_pos[0:2]))/c*1e9 # Save the delay from the object of interest
                            obj_params[k, 2] = np.dot(obj_vel, tx_locations - obj_pos)/np.linalg.norm(tx_locations - obj_pos)*f_c/c + \
                                               np.dot(obj_vel, rx_locations - obj_pos)/np.linalg.norm(rx_locations - obj_pos)*f_c/c
                            obj_params[k, 3] = 1 # Save the higher order visibility of the object
                            print("{}th path of {}th iteration is a detectable higher order reflection from the target".format(int(l), k))
                        else:
                            print("{}th path of {}th iteration is an undetectable higher order reflection from the target".format(int(l), k))
                        
                    break
    
            parameters[k, 0, i_path] = np.array(paths.phi_r[0,0,0,0,int(l)])
            parameters[k, 1, i_path] = Doppler
            parameters[k, 2, i_path] = np.array(paths._tau[0,0,0,0,int(l)])
            
            parameters[k, 3, i_path] = a_paths[int(l)]

            i_path = i_path + 1
    
    print("The location of object in gNB frame of reference is [{}, {}]".format(rx_locations[0]-obj_pos[0], rx_locations[1]-obj_pos[1]))
    print("{}th iteration is done".format(k))
    
    positions[k, 0] = rx_locations[0]-obj_pos[0]
    positions[k, 1] = rx_locations[1]-obj_pos[1]
    
    obj_pos = obj_pos + Te*obj_vel
    scene.get('elm__4').position = mi.Point3f(obj_pos)


np.save('/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Trajectory_1/ISAC_ray_tracing_chan.npy', parameters)
np.save('/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Trajectory_1/ISAC_locations.npy', positions)
np.save('/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Trajectory_1/obj_params.npy', obj_params)
```

```python
### Generating Trajectory 3 ###

# Locations of the Transmitter and receiver
rx_locations = np.array([-13, 20, 1]) 
tx_locations = np.array([-13, 0, 1])

# - - - - - - - - - START: Trajectory points follow sinusoidal wave - - - - - - - - - #

# Parameters
A = 3.0           # Amplitude (m)
wavelength = 5.0  # Wavelength (m)
ds = 0.03         # Fixed distance along curve (m)
K = 500 # No. of times the simulation will run
k = 2 * np.pi / wavelength

y_points = [0.0]
x_points = [0.0]

current_y = 0.0

for _ in range(K): # One extra point is calculated here to calculate the velocity later
    # Calculate dy from ds based on current slope
    dx_dy = A * k * np.cos(k * current_y)
    dy = ds / np.sqrt(1 + dx_dy**2)
    
    current_y += dy
    current_x = A * np.sin(k * current_y)
    
    y_points.append(current_y)
    x_points.append(current_x)

points = np.column_stack((x_points, y_points, np.zeros(K+1))) + np.tile(np.array([[-6, 8, 0.9]]), [K+1, 1])

# - - - - - - - - - END - - - - - - - - - #

init_obj_pos = np.array([100, 200, 0.9]) # Initial position of the object in static environment

# Initial 30 points will be static points so the object is not available in the environment
points[0:30, :] = np.tile(np.reshape(init_obj_pos, [1, 3]), [30, 1])

obj_dim = np.array([0.25, 0.5, 1.8]) # Object dimension
obj_ori = np.array([0, 0, 0])

Te = 0.02 # Seconds between successive simulation runs

f_c = 3.6192e9
c = 299792458
max_depth = 4
L_paths = 100

# Carrier frequency
scene.frequency = f_c

parameters = np.zeros([K, 4, L_paths]) # Save the channel parameters in a numpy array for being used by ISAC simulator
positions = np.zeros([K, 2]) # Save the position of the object
obj_params = np.zeros([K, 4]) # Save the true AoA, ToF and visibility of the object to the transmitter
detect_snr_thres = 30 # This is the largest ratio of LoS to target path power that makes it a GT detection

# Load the Sionna scene 
scene = load_scene("/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Blender/Outdoor.xml", merge_shapes=False)

# Configure antenna array for all transmitters
scene.tx_array = PlanarArray(num_rows=1,
                             num_cols=1,
                             vertical_spacing=0.5,
                             horizontal_spacing=0.5,
                             pattern="iso",
                             polarization="V")

# Configure antenna array for all receivers
scene.rx_array = PlanarArray(num_rows=1,
                             num_cols=1,
                             vertical_spacing=0.5,
                             horizontal_spacing=0.5,
                             pattern="iso",
                             polarization="VH")

tx = Transmitter(name="tx",
                 position=tx_locations,
                 display_radius=3)

scene.add(tx)

rx = Receiver(name="rx",
              position=rx_locations,
              display_radius=3)

scene.add(rx)

# Instantiate a path solver
# The same path solver can be used with multiple scenes
p_solver  = PathSolver()
scene.get('elm__4').orientation = mi.Point3f(obj_ori)
scene.get('elm__4').radio_material.scattering_coefficient = 1


for k in range(0, K):

    obj_pos = points[k, :]
    obj_vel = (points[k+1, :] - points[k, :])/Te
    scene.get('elm__4').position = mi.Point3f(obj_pos)
    
    # Compute propagation paths
    paths = p_solver(scene,
                     max_depth=max_depth,
                     #max_num_paths_per_src=3,
                     los=True,
                     specular_reflection=True,
                     diffuse_reflection=True,
                     diffraction=True,
                     edge_diffraction=False,
                     refraction=True,
                     synthetic_array=False,
                     seed=41)
    
    poi = np.array(paths.vertices[:, 0, 0, 0, :, :]) # Point of interactions
    toi = np.array(paths.interactions[:, 0, 0, 0, 0, :]) # Type of interactions
    L = np.shape(poi)[2]
    a_v = np.array(paths._a_real[0,0,0,0,:]) + 1j*np.array(paths._a_imag[0,0,0,0,:])
    a_h = np.array(paths._a_real[0,1,0,0,:]) + 1j*np.array(paths._a_imag[0,1,0,0,:])
    a_paths = np.abs(a_v)**2 + np.abs(a_h)**2

    sort_path_idx = np.argsort(a_paths)[::-1]
    los_power = np.max(a_paths)
    i_path = 0
    
    for l in sort_path_idx:
        Doppler = 0
        if (paths.phi_r[0,0,0,0,int(l)] < 0) and (i_path < L_paths): # We are ignoring all the paths that are received from behind of the BS array and also ignoring any path after the strongest 'L_paths'
            
            for order in range(0, max_depth): # Check across the orders of interactions for interactions with the target
                if check_inside(poi[order,0,int(l),:], obj_pos, obj_ori, obj_dim):
                    obj_point = np.array(poi[order, 0, int(l), :]) # Use this as the point of interaction when calculating the Doppler
                    
                    if (order == 0) and toi[1, int(l)] == 0: #(poi[1,0,l,0] == 0 and poi[1,0,l,1] == 0 and poi[1,0,l,2] == 0): # This cannot be used because for some reason, vertices[] array is filled with non-zero coordinates even when interactions is zero for that order of that path.
                        Doppler = np.dot(obj_vel, tx_locations - obj_point)/np.linalg.norm(tx_locations - obj_point)*f_c/c + \
                                  np.dot(obj_vel, rx_locations - obj_point)/np.linalg.norm(rx_locations - obj_point)*f_c/c  # Doppler shift due to the relative motion of the object w.r.t. the BS and UE
                        
                        if a_paths[l]/los_power >= 10**(-0.1*detect_snr_thres):
                            obj_params[k, 0] = 180/np.pi*np.atan2(obj_pos[1]-rx_locations[1], obj_pos[0]-rx_locations[0]) # Save the AoA from the object of interest
                            obj_params[k, 1] = (np.linalg.norm(rx_locations[0:2]-obj_pos[0:2]) + np.linalg.norm(tx_locations[0:2]-obj_pos[0:2]))/c*1e9 # Save the delay from the object of interest
                            obj_params[k, 2] = np.dot(obj_vel, tx_locations - obj_pos)/np.linalg.norm(tx_locations - obj_pos)*f_c/c + \
                                               np.dot(obj_vel, rx_locations - obj_pos)/np.linalg.norm(rx_locations - obj_pos)*f_c/c
                            obj_params[k, 3] = 1 # Save the direct visibility of the object
                            print("{}th path of {}th iteration is a detectable first order reflection from the target".format(int(l), k))
                        # else:
                            # print("{}th path of {}th iteration is an undetectable first order reflection from the target".format(int(l), k))

                    else:
                        if order == 0:
                            first_end = tx_locations
                            second_end = np.array(poi[1, 0, int(l), :])
                        elif order == max_depth-1:
                            first_end = np.array(poi[max_depth-2, 0, int(l), :])
                            second_end = rx_locations
                        else:
                            first_end = np.array(poi[order-1, 0, int(l), :])
                            second_end = np.array(poi[order+1, 0, int(l), :])

                        Doppler = np.dot(obj_vel, first_end - obj_point)/np.linalg.norm(first_end - obj_point)*f_c/c + \
                                  np.dot(obj_vel, second_end - obj_point)/np.linalg.norm(second_end - obj_point)*f_c/c  # Doppler shift due to the relative motion of the object w.r.t. the environment
                        
                        if a_paths[l]/los_power >= 10**(-0.1*detect_snr_thres):
                            obj_params[k, 0] = 180/np.pi*np.atan2(obj_pos[1]-rx_locations[1], obj_pos[0]-rx_locations[0]) # Save the AoA from the object of interest
                            obj_params[k, 1] = (np.linalg.norm(rx_locations[0:2]-obj_pos[0:2]) + np.linalg.norm(tx_locations[0:2]-obj_pos[0:2]))/c*1e9 # Save the delay from the object of interest
                            obj_params[k, 2] = np.dot(obj_vel, tx_locations - obj_pos)/np.linalg.norm(tx_locations - obj_pos)*f_c/c + \
                                               np.dot(obj_vel, rx_locations - obj_pos)/np.linalg.norm(rx_locations - obj_pos)*f_c/c
                            obj_params[k, 3] = 1 # Save the higher order visibility of the object
                            print("{}th path of {}th iteration is a detectable higher order reflection from the target".format(int(l), k))
                        # else:
                            # print("{}th path of {}th iteration is an undetectable higher order reflection from the target".format(int(l), k))
                        
                    break
    
            parameters[k, 0, i_path] = np.array(paths.phi_r[0,0,0,0,int(l)])
            parameters[k, 1, i_path] = Doppler
            parameters[k, 2, i_path] = np.array(paths._tau[0,0,0,0,int(l)])
            
            parameters[k, 3, i_path] = a_paths[int(l)]

            i_path = i_path + 1
    
    print("The location of object in gNB frame of reference is [{}, {}]".format(rx_locations[0]-obj_pos[0], rx_locations[1]-obj_pos[1]))
    print("{}th iteration is done".format(k))
    
    positions[k, 0] = rx_locations[0]-obj_pos[0]
    positions[k, 1] = rx_locations[1]-obj_pos[1]
    

np.save('/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Trajectory_3/ISAC_ray_tracing_chan.npy', parameters)
np.save('/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Trajectory_3/ISAC_locations.npy', positions)
np.save('/home/bkumar2/Documents/Research/ISAC/Python_Scripts_CellSense2.0/Simulation/Datasets/Outdoor_scenario/Trajectory_3/obj_params.npy', obj_params)
```
